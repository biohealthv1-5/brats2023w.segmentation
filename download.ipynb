```python
import hashlib
import os

def verify_data(file_path, expected_hash):
    # Complexity: O(N) where N is file size
    # Performance: Ensures data integrity before training
    hasher = hashlib.md5()
    with open(file_path, 'rb') as f:
        for chunk in iter(lambda: f.read(4096), b""):
            hasher.update(chunk)
    
    current_hash = hasher.hexdigest()
    if current_hash == expected_hash:
        return True
    else:
        return False

# Synapse 클라이언트는 내부적으로 이 과정을 수행하므로, 
# 명령어를 재실행했을 때 "skipping"이 뜨면 안전하다고 판단할 수 있습니다.
```
